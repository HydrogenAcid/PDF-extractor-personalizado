```python
import numpy as np #importar la libreria de numpy con un alias
```

```python
#*  Beta_Series.ipynb
# Documentado por: Braulio
#*  Descripcion: Este programa simula una serie de Fourier con coeficientes que siguen una ley de potencia, es decir, el valor de cada coeficiente se obtiene multiplicando un número aleatorio por una función de potencia que depende del índice del coeficiente. La función Beta_Simulation toma dos parámetros: B, que controla la tasa de decaimiento de los coeficientes, y N, que es el número de coeficientes a generar. El programa utiliza la transformada rápida de Fourier (FFT) para generar los coeficientes y luego aplica la transformada inversa para obtener la serie simulada.
def Beta_Simulation(B, N): #declara una funcion que recibe B,N como entradas llamada
    X = np.random.normal(0,1,N) #declara una variable "X" la cual contiene un arreglo de N numeros aleatorios con una distribucion gaussiana normal con una media de 0 y std de 1
    Y = np.fft.fft(X) #declara una variable "Y" dependiente de la anterior y de la misma dimension la cual contiene la transformada de Fourier de X
    n = len(Y) #int n que vale la longitud de "Y"
    f=[0 for i in range(n)] #f contiene un arreglo de ceros de la misma dimension que Y,n
    h=1 #int h con un valor inicial de 1
    #bucle espejo --> [0,x,x,x,x,x,x,x,x,x,x,0] 
    while h < n/2: #bucle donde la condicion de paro es cuando h es menor a n/2
        aux = Y[h]*((float(h)/float(n))**(-B/2.0)) #aux es igual a Y[h] multiplicado por el resultado de la division de h entre n elevado a la potencia de -B/2.0 el cual se va iterar con base en la linea anterior
        f[h] = aux #asigna valores del segundo indice hasta el centro del arreglo del lado derecho de f con el valor de aux
        f[n-h-1] = aux#asigna valores del lado izquierdo hasta el centro
        h=h+1#h se va incrementar en 1 hasta llegar a n/2 o a la mitad de su dimension 
    g=np.fft.ifft(f)#g= a la Inversa de la transformada rapida de F. de f
    return g.real#retorna en este caso, la parte real de g
```
